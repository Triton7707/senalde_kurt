```python
import pandas as pd
```

```python
df = pd.read_csv('student_depression_dataset.csv')
```

```python
df.columns = df.columns.str.lower().str.replace(' ', '_')
```

```python
df = df.drop_duplicates()
```

```python
for col in df.columns:
    if df[col].dtype in ["int64", "float64"]:
        df[col] = df[col].fillna(df[col].median())
    else:
        df[col] = df[col].fillna("unknown")
```

```python
for col in df.select_dtypes(include="object").columns:
    df[col] = df[col].str.strip().str.lower()
```

```python
if "gender" in df.columns:
    df["gender"] = df["gender"].replace({
        "m": "male",
        "f": "female"
    })
```

```python
numeric_cols = [
    "age", "cgpa", "sleep_duration",
    "academic_pressure", "study_satisfaction"
]

for col in numeric_cols:
    if col in df.columns:
        df[col] = pd.to_numeric(df[col], errors="coerce")
```

```python
if "age" in df.columns:
    df["age"] = df["age"].clip(lower=10, upper=60)

if "cgpa" in df.columns:
    # Adjust depending on your dataset scale (0–10 is safer than 0–5)
    df["cgpa"] = df["cgpa"].clip(lower=0, upper=10)
```

```python
df.to_csv("cleaned_dataset.csv", index=False)

print("Cleaning complete. File saved as 'cleaned_dataset.csv'")
```
